```python
print("Categorical features:")
print(df.select_dtypes(include = "object").columns.tolist())
```

```python
product_sales = df["product_line"].value_counts()
print("Highst product of sales:")
print(product_sales)
```

```python
branch_sales = df.groupby("branch")["revenue"].sum().sort_values()

print("Hights branch of sales:")
print(branch_sales)
```

```python
ave_rating = df["rating"].mean()
print("Average Customer Rating:")
print(ave_rating)
```

```python
Payment_method = df["payment_method"].value_counts()

print("Most used payment method")
print(Payment_method)
```

```python
plt.figure(figsize = (5,10))
product_sales.plot(kind = "bar")
plt.title("Number of sales by products")
plt.xlabel("Product")
plt.ylabel("Number of sales")
plt.xticks(rotation = 45)

plt.show()
```
